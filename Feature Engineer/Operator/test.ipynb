```python
# 对于基于单特征进行增强的算子，全量进行测试
from pathlib import Path

CHUNK_SIZE = 200

# ops_list = [
#     ("ts_macd", {"func": ops.ts_macd, "args": (df, cols), "kwargs": {"short_window": 12, "long_window": 26, "signal_window": 9, "group_col": "stock_id"}}),
#     ("ts_rolling_rank", {"func": ops.ts_rolling_rank, "args": (df, cols), "kwargs": {"windows": [5, 10, 20], "group_col": "stock_id"}}),
#     ("ts_regbeta", {"func": ops.ts_regbeta, "args": (df, cols), "kwargs": {"target_col": "wap", "windows": [5, 10], "group_col": "stock_id"}}),
#     ("ts_bollinger_bands", {"func": ops.ts_bollinger_bands, "args": (df, cols), "kwargs": {"window": 20, "num_sd": 2, "group_col": "stock_id"}}),

#     ("cs_deviation_from_mean", {"func": ops.cs_deviation_from_mean, "args": (df, cols), "kwargs": {"group_col": "time_id"}}),
#     ("cs_decompose_sigmoid", {"func": ops.cs_decompose_sigmoid, "args": (df, cols), "kwargs": {"n": 4, "group_col": "time_id"}}),
#     ("cs_decompose_tanh", {"func": ops.cs_decompose_tanh, "args": (df, cols), "kwargs": {"n": 4, "group_col": "time_id"}}),
# ]


ops_list = [
    ("ts_rsi", {
        "func": ops.ts_rsi,
        "args": (df, cols),
        "kwargs": {"period": 14, "group_col": "stock_id"},
    }),
    ("ts_ema", {
        "func": ops.ts_ema,
        "args": (df, cols),
        "kwargs": {"spans": [5, 10, 20], "group_col": "stock_id"},
    }),
    ("ts_momentum", {
        "func": ops.ts_momentum,  # 等价 ts_diff
        "args": (df, cols),
        "kwargs": {"periods": [1, 5, 10], "group_col": "stock_id"},
    }),
    ("ts_mean", {
        "func": ops.ts_mean,
        "args": (df, cols),
        "kwargs": {"windows": [5, 10, 20], "group_col": "stock_id"},
    }),
]

output_dir = Path("data/output")
output_dir.mkdir(parents=True, exist_ok=True)

all_results: list[pd.DataFrame] = []

for op_name, op_info in ops_list:
    print(f'开始处理算子 {op_name} ...')

    op_args = op_info.get("args", ())
    op_kwargs = op_info.get("kwargs", {})
    try:
        op_df = op_info["func"](*op_args, **op_kwargs)
    except Exception as e:
        print(f'算子 {op_name} 计算失败: {str(e)}，跳过处理')
        del op_args, op_kwargs
        continue

    new_df = op_df
    del op_df, op_args, op_kwargs

    factor_cols = [col for col in new_df.columns if col not in DEFAULT_EXCLUDE_COLS]
    if not factor_cols:
        print(f'{op_name}: 无可用因子列，跳过。')
        del new_df
        continue

    factor_chunks = [factor_cols[i:i + CHUNK_SIZE] for i in range(0, len(factor_cols), CHUNK_SIZE)]
    total_chunks = len(factor_chunks)

    final_save_cols: list[str] = []
    total_selected_count = 0
    total_remaining_count = 0

    for chunk_idx, chunk_cols in enumerate(factor_chunks, start=1):
        print(f"{op_name}: 开始处理第 {chunk_idx}/{total_chunks} 个 chunk，共 {len(chunk_cols)} 个因子")

        backtest_required_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(chunk_cols)))
        df_backtest = new_df.select(backtest_required_cols)

        try:
            result = backtest(
                df_backtest,
                chunk_cols,
                threshold=0.03,
                start_date=0,
                end_date=400,
                check_inf=True,
            )
        except Exception as e:
            print(f'{op_name} chunk {chunk_idx} 回测失败: {str(e)}，跳过该 chunk')
            del df_backtest
            continue

        del df_backtest

        if isinstance(result, pl.DataFrame):
            result_pd = result.to_pandas()
        else:
            result_pd = result.copy()
        del result

        all_results.append(result_pd)

        selected_cols = result_pd.loc[result_pd['ic'].abs() >= 0.03, 'factor'].tolist()
        selected_cols = list(set([col for col in selected_cols if col in new_df.columns and col in chunk_cols]))

        if not selected_cols:
            print(f'{op_name} chunk {chunk_idx}: 没有通过 IC 筛选的因子，跳过。')
            del result_pd
            continue

        total_selected_count += len(selected_cols)

        filter_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(selected_cols)))
        df_filtered = new_df.select(filter_cols)
        remaining_cols, removed_cols = filter_correlated_factors(
            df=df_filtered,
            cols=selected_cols,
            ic_df=result_pd,
            corr_threshold=0.8,
            factor_col="factor",
            ic_col="ic",
            method="corr_csall"
        )

        save_cols = [col for col in remaining_cols if col not in DEFAULT_COLS]
        if not save_cols:
            print(f'{op_name} chunk {chunk_idx}: 相关性筛选后无可用因子，跳过。')
            del df_filtered, result_pd
            continue

        final_save_cols.extend(save_cols)
        total_remaining_count += len(save_cols)

        del df_filtered, result_pd

    if not final_save_cols:
        print(f'{op_name}: 所有 chunk 在筛选后均无可用因子，跳过保存。')
        del new_df
        continue

    unique_save_cols = list(dict.fromkeys(final_save_cols))
    final_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(unique_save_cols)))
    df_final = new_df.select(final_cols)
    output_path = output_dir / f'{op_name}.parquet'
    df_final.write_parquet(str(output_path))

    print(
        f'{op_name}: '
        f'通过 IC 的因子 {total_selected_count} 个，'
        f'相关性筛后 {total_remaining_count} 个，'
        f'最终保存 {len(unique_save_cols)} 个，'
        f'已保存到 {output_path}'
    )

    del new_df, df_final

print("所有算子处理完成")
```

```python
# 分组筛选 + 组内/组间组合，并仿照单特征算子流程回测与保存

CHUNK_SIZE=200

group_defs = {
    "liquidity": cols_liquidity,
    "imbalance": cols_imbalance,
    "deviation": cols_deviation,
    "trend_impact": cols_trend_impact,
}

group_filtered_cols: dict[str, list[str]] = {}
for group_name, base_cols in group_defs.items():
    candidates = [col for col in base_cols if col in df.columns]
    if not candidates:
        print(f"{group_name}: 无可用列，跳过。")
        continue

    df_filtered = df.select(list(dict.fromkeys(list(DEFAULT_COLS) + candidates)))
    ic_subset = ic_df[ic_df["factor"].isin(candidates)]

    # remaining_cols, removed_cols = filter_correlated_factors(
    #     df=df_filtered,
    #     cols=candidates,
    #     ic_df=ic_subset,
    #     corr_threshold=0.9,
    #     factor_col="factor",
    #     ic_col="ic",
    #     method="corr_csall"
    # )

    # group_filtered_cols[group_name] = remaining_cols
    # print(f"{group_name}: 保留 {len(remaining_cols)} 列，剔除 {len(removed_cols)} 列")

    group_filtered_cols[group_name] = candidates

# 组间：生成列对，再做交叉与除法
col_pairs: list[tuple[str, str]] = []
group_names = list(group_filtered_cols.keys())
for i in range(len(group_names)):
    for j in range(i + 1, len(group_names)):
        for c1 in group_filtered_cols[group_names[i]]:
            for c2 in group_filtered_cols[group_names[j]]:
                col_pairs.append((c1, c2))

col_pairs = list(dict.fromkeys(col_pairs))
pair_cols = list(dict.fromkeys([c for pair in col_pairs for c in pair]))


def _build_cs_zscore_add():
    op_df = df
    for group_name, group_cols in group_filtered_cols.items():
        if len(group_cols) < 2:
            print(f"{group_name}: 组内可用列不足 2 个，跳过 cs_zscore_add。")
            continue
        op_df = ops.cs_zscore_add(op_df, group_cols, group_col="time_id")
    return op_df


def _build_cs_cross_feature():
    return ops.cs_cross_feature(df, cols=pair_cols, col_pairs=col_pairs)


def _build_cs_divide():
    return ops.cs_divide(df, cols=pair_cols, col_pairs=col_pairs)


group_ops = [
    # ("group_cs_zscore_add", _build_cs_zscore_add),
    ("group_cs_cross_feature", _build_cs_cross_feature),
    # ("group_cs_divide", _build_cs_divide),
]

output_dir = Path("data/output")
output_dir.mkdir(parents=True, exist_ok=True)

all_results_group: list[pd.DataFrame] = []


def run_group_op(op_name: str, op_builder):
    print(f"开始处理算子 {op_name} ...")
    op_df = op_builder()

    factor_cols = [
        col for col in op_df.columns
        if col not in DEFAULT_EXCLUDE_COLS
    ]
    if not factor_cols:
        print(f"{op_name}: 无新增因子列，跳过。")
        return

    factor_chunks = [
        factor_cols[i:i + CHUNK_SIZE] for i in range(0, len(factor_cols), CHUNK_SIZE)
    ]
    total_chunks = len(factor_chunks)

    final_save_cols: list[str] = []
    total_selected_count = 0
    total_remaining_count = 0

    for chunk_idx, chunk_cols in enumerate(factor_chunks, start=1):
        print(f"{op_name}: 开始处理第 {chunk_idx}/{total_chunks} 个 chunk，共 {len(chunk_cols)} 个因子")

        backtest_required_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(chunk_cols)))
        df_backtest = op_df.select(backtest_required_cols)

        try:
            result = backtest(
                df_backtest,
                chunk_cols,
                threshold=0.03,
                start_date=0,
                end_date=400,
                check_inf=True,
            )
        except Exception as e:
            print(f"{op_name} chunk {chunk_idx} 回测失败: {str(e)}，跳过该 chunk")
            del df_backtest
            continue

        del df_backtest

        if isinstance(result, pl.DataFrame):
            result_pd = result.to_pandas()
        else:
            result_pd = result.copy()
        del result

        all_results_group.append(result_pd)

        selected_cols = result_pd.loc[result_pd["ic"].abs() >= 0.03, "factor"].tolist()
        selected_cols = list(set([col for col in selected_cols if col in op_df.columns and col in chunk_cols]))

        if not selected_cols:
            print(f"{op_name} chunk {chunk_idx}: 没有通过 IC 筛选的因子，跳过。")
            del result_pd
            continue

        total_selected_count += len(selected_cols)

        filter_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(selected_cols)))
        df_filtered = op_df.select(filter_cols)
        remaining_cols, removed_cols = filter_correlated_factors(
            df=df_filtered,
            cols=selected_cols,
            ic_df=result_pd,
            corr_threshold=0.8,
            factor_col="factor",
            ic_col="ic",
            method="corr_csall"
        )

        save_cols = [col for col in remaining_cols if col not in DEFAULT_COLS]
        if not save_cols:
            print(f"{op_name} chunk {chunk_idx}: 相关性筛选后无可用因子，跳过。")
            del df_filtered, result_pd
            continue

        final_save_cols.extend(save_cols)
        total_remaining_count += len(save_cols)

        del df_filtered, result_pd

    if not final_save_cols:
        print(f"{op_name}: 所有 chunk 在筛选后均无可用因子，跳过保存。")
        return

    unique_save_cols = list(dict.fromkeys(final_save_cols))
    final_cols = list(dict.fromkeys(list(DEFAULT_COLS) + list(unique_save_cols)))
    df_final = op_df.select(final_cols)
    output_path = output_dir / f"{op_name}.parquet"
    df_final.write_parquet(str(output_path))

    print(
        f"{op_name}: "
        f"通过 IC 的因子 {total_selected_count} 个，"
        f"相关性筛后 {total_remaining_count} 个，"
        f"最终保存 {len(unique_save_cols)} 个，"
        f"已保存到 {output_path}"
    )


for op_name, op_builder in group_ops:
    run_group_op(op_name, op_builder)

print("分组组合算子处理完成")
```
